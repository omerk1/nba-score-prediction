```python
# Quick exploration if database exists
import sqlite3

if data_path.exists():
    conn = sqlite3.connect(f"file:{data_path}?mode=ro", uri=True)
    tables = pd.read_sql("SELECT name FROM sqlite_master WHERE type='table';", conn)
    print("Available tables:")
    for i, table in enumerate(tables['name'], 1):
        n = pd.read_sql(f"SELECT COUNT(*) AS n FROM [{table}]", conn)['n'][0]
        print(f"  {i}. {table} ({n:,} rows)")
    span = pd.read_sql("SELECT MIN(game_date) AS lo, MAX(game_date) AS hi FROM game", conn)
    print(f"\ngame spans {span['lo'][0]} to {span['hi'][0]}")
    conn.close()
```
